# 例2-23  在一张图片中添加opencv logo

```python
# step1   读取原始图片
import cv2
import matplotlib.pyplot as plt
import numpy as np

grassland = cv2.imread('/home/aistudio/work/grassland.png')
grassland = cv2.cvtColor(grassland, cv2.COLOR_BGR2RGB)
plt.imshow(grassland)
plt.show()
print(grassland.shape)

logo = cv2.imread('/home/aistudio/work/opencv-logo-white.png')
logo = cv2.cvtColor(logo, cv2.COLOR_BGR2RGB)
plt.imshow(logo)
plt.show()
print(grassland.shape)
print(logo.shape)
```

```python
# step2  enlarge  logo image to original image size 
origin_image = grassland 
logo_image =  logo 
logo_image_enlarge = np.ones(origin_image.shape,dtype = np.uint8) * 255
plt.imshow(logo_image_enlarge)
plt.show()
logo_h = logo_image.shape[0]
logo_w  = logo_image.shape[1]
logo_c = logo_image.shape[2]
logo_image_enlarge[0:logo_h,0:logo_w,:] =logo_image[:,:,:] 
plt.imshow(logo_image_enlarge)
plt.show()
logo_image_enlarge_gray = cv2.cvtColor(logo_image_enlarge,cv2.COLOR_RGB2GRAY)
plt.imshow(logo_image_enlarge_gray,cmap = 'gray')
plt.show()
print(logo_image_enlarge_gray.shape)
logo_image_enlarge_mask = logo_image_enlarge_gray>200
plt.imshow(logo_image_enlarge_mask,cmap = 'gray')
plt.show()
logo_mask = np.zeros(logo_image_enlarge_mask.shape,dtype = np.uint8)
logo_mask[logo_image_enlarge_mask==False] = 255
plt.imshow(logo_mask,cmap= 'gray')
plt.show()
logo_mask_rgb = np.zeros(origin_image.shape,dtype = np.uint8)
logo_mask_rgb[:,:,0] = logo_mask
logo_mask_rgb[:,:,1] = logo_mask
logo_mask_rgb[:,:,2] = logo_mask

logo_mask_rgb_neg = np.bitwise_not(logo_mask_rgb)
plt.imshow(logo_mask_rgb_neg)
plt.show()

print(logo_mask_rgb_neg)
origin_masked = np.bitwise_and(origin_image, logo_mask_rgb_neg) 
plt.imshow(origin_masked)
plt.show()  

logo_masked = np.bitwise_and(logo_image_enlarge,logo_mask_rgb)
plt.imshow(logo_masked)
plt.show()

logoed_image = origin_masked + logo_masked 
plt.imshow(logoed_image)
plt.show()




print(logo_image_enlarge_mask)
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

grassland = cv2.imread('/home/aistudio/work/grassland.png')
grassland = cv2.cvtColor(grassland, cv2.COLOR_BGR2RGB)
plt.imshow(grassland)
plt.show()



logo = cv2.imread('/home/aistudio/work/opencv-logo-white.png')
logo = cv2.cvtColor(logo, cv2.COLOR_BGR2RGB)
plt.imshow(logo)
plt.show()
print(grassland.shape)
print(logo.shape)

mask_white = 255*np.ones(grassland.shape,dtype=np.uint8)
print(mask_white.dtype)
plt.imshow(mask_white)
plt.show()

roi = mask_white[0:0+268,400:400+268,:] #region of interest
roi[:,:,:] = logo[:,:,:]
plt.imshow(mask_white)
plt.show()

opencv_color = mask_white.copy()
plt.imshow(opencv_color)
plt.title('opencv color')
plt.show()

mask_white_gray = (mask_white[:,:,0].astype(np.float64) + mask_white[:,:,1].astype(np.float64)
                   + mask_white[:,:,2].astype(np.float64))/3.0
mask_white_gray = mask_white_gray.astype(np.uint8)
print(mask_white_gray.dtype)
print(mask_white_gray)
plt.imshow(mask_white_gray,cmap='gray')
plt.show()

mask_white_gray[mask_white_gray<128] = 0
mask_white_gray[mask_white_gray>=128]=255 
plt.imshow(mask_white_gray,cmap='gray')
plt.title('gray')
plt.show()
print(mask_white_gray)


mask_inv = mask_white_gray.copy()
mask = mask_white_gray.copy()
mask = np.bitwise_not(mask)

mask_color = np.zeros(grassland.shape,dtype= np.uint8)
mask_color[:,:,0] = mask
mask_color[:,:,1] = mask
mask_color[:,:,2]  = mask

mask_color_inv = mask_color.copy()
mask_color_inv =np.bitwise_not(mask_color_inv)

plt.imshow(mask_inv,cmap='gray')
plt.title('mask_inv')
plt.show()
print(mask_inv)
plt.imshow(mask,cmap='gray')
plt.title('mask')
plt.show()
print(mask)

plt.imshow(mask_color)
plt.title('mask_color')
plt.show()
print(mask_color)
plt.imshow(mask_color_inv)
plt.title('mask_color_inv')
plt.show()
print(mask_color_inv)


opencv_masked = opencv_color & mask_color
plt.imshow(opencv_masked)
plt.title('opencv masked')
plt.show()

grassland_masked = grassland & mask_color_inv
plt.imshow(grassland_masked)
plt.title('grasssland masked')
plt.show()

grassland_add_opecvlogo = grassland_masked + opencv_masked
plt.imshow(grassland_add_opecvlogo)
plt.title('grassland add opencvlogo')
plt.show()

cv2.cvtColor(grassland_add_opecvlogo, cv2.COLOR_RGB2BGR,grassland_add_opecvlogo)
cv2.imwrite('/home/aistudio/work/grassland_add_opencvlogo.jpg',grassland_add_opecvlogo)


```
