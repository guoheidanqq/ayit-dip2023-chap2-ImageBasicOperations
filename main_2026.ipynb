# Example 2-10

## resize   cv2 总是WHC
## numpy    HWC
## pytorch   CHW

```python
import cv2 
import matplotlib.pyplot as plt
image1 = cv2.imread('/home/aistudio/ayit-dip2023-chap2-ImageBasicOperations/IMG_20260915_074514.jpg')
image2 = cv2.imread('/home/aistudio/work/moon_resize.png')
image2 = cv2.resize(image2,dsize = (3072,4096))

print(image1.shape)
print(image2.shape)
print(image1.dtype)
print(image2.dtype)
print(type(image1))
print(type(image2))

image3 = image1 + image2 
image3_cvadd= cv2.add(image1,image2)
image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
image3_rgb = cv2.cvtColor(image3, cv2.COLOR_BGR2RGB)
image3_rgb_cvadd = cv2.cvtColor(image3_cvadd,cv2.COLOR_BGR2RGB)
plt.imshow(image1_rgb,cmap='gray', vmin=0, vmax=255)
plt.show()
plt.imshow(image2_rgb,cmap='gray', vmin=0, vmax=255)
plt.show()
plt.imshow(image3_rgb,cmap='gray', vmin=0, vmax=255)
plt.show()
plt.imshow(image3_rgb_cvadd,cmap='gray', vmin=0, vmax=255)
plt.show()
```
